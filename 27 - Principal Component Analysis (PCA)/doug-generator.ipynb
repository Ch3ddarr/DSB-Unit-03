```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
doug = plt.imread('doug.jpg')
doug = np.mean(doug, axis=2)
```

```python
u, s, vh = np.linalg.svd(doug)
```

```python
plt.imshow(1 - doug, cmap='Greys')
plt.tight_layout()
plt.savefig('doug-full.png')
```

```python
def plot_doug(ax, k, t):
    ax.imshow(1 - u[:, :k] @ np.diag(s)[:k, :k] @ vh[:k, :], cmap='Greys')
    ax.set_title(t)
```

```python
fig, ax = plt.subplots(1, 1)
plot_doug(ax, 75, 'Doug?!')
```

```python
fig, axs = plt.subplots(2, 3, figsize=(20, 12))
plot_doug(axs[0][0], 900, 'Full image ($k = 900$)')
plot_doug(axs[0][1], 100, '$k = 100$')
plot_doug(axs[0][2], 50, '$k = 50$')
plot_doug(axs[1][0], 20, '$k = 20$')
plot_doug(axs[1][1], 10, '$k = 10$')
plot_doug(axs[1][2], 5, '$k = 5$')
plt.tight_layout()
plt.savefig('doug-pca.png')
```
